# Used Car Depreciation Analysis — Data Analysis

This notebook analyzes vehicle depreciation and value retention across four powertrain categories using the processed dataset generated during the data preparation phase.

The analysis aims to provide preliminary valuation benchmarks to support used-car dealership trade-in assessments and inventory acquisition decisions.

## Analysis Objectives

- Compare value retention across Gasoline, HEV, PHEV, and BEV vehicles.
- Evaluate depreciation patterns across vehicle age intervals.
- Compare value retention at approximately 3, 5, and 10 years.
- Illustrate the potential financial implications of depreciation differences.
- Develop actionable business recommendations for dealership acquisition pricing.


## Import Libraries and Load Dataset

Import Pandas and load the processed depreciation
dataset generated during the data preparation phase.


```python
import pandas as pd
import matplotlib.pyplot as plt

df_depreciation_processed = pd.read_csv("../data/processed/depreciation_processed.csv")

df_depreciation_processed.head()
```

## Validate Imported Dataset

Verify that the imported dataset maintains the expected
dimensions, column names, data types, and completeness
established during the data preparation phase.


```python
df_depreciation_processed.shape
```

## Compare Value Retention Across Powertrains

Reshape the dataset to compare vehicle value retention
across the four powertrain categories at equivalent
vehicle age intervals.


```python
df_retention_pivot = df_depreciation_processed.pivot(
    index="age_years", columns="powertrain", values="retention_pct"
)

df_retention_pivot.head()
```

```python
df_retention_pivot.shape
```

## Visualize Value Retention Trends

Create a line chart to compare vehicle value retention
across the four powertrain categories as vehicles age.

Identify depreciation patterns, differences in value
retention, and potential implications for dealership
inventory acquisition decisions.


```python
df_retention_pivot.plot(
    figsize=(12, 6),
    title="Used Vehicle Value Retention by Powertrain",
    xlabel="Vehicle Age (Years)",
    ylabel="Value Retention (%)",
    grid=True,
)

plt.savefig(
    "../visualizations/value_retention_trends.png",
    dpi=300,
    bbox_inches="tight",
)

plt.show()
```

### Initial Observations

- HEVs generally demonstrate stronger value retention
  during the first three years of vehicle age.

- Gasoline and HEV vehicles follow similar depreciation
  trends. Gasoline vehicles generally retain more value
  during the middle years before both categories converge
  around years 14–15.

- PHEVs and BEVs exhibit similar depreciation patterns,
  although PHEVs generally demonstrate stronger value
  retention during later years.

- PHEVs exhibit an unexpected increase in value retention
  around year five, which warrants further investigation.

These preliminary observations are further evaluated
through quantitative comparisons at selected vehicle
age milestones.


## Value Retention at Key Vehicle Ages

Compare median value retention across the four powertrain categories at approximately 3, 5, and 10 years.

The available age intervals of 3.125, 5.125, and 10.125 years are used consistently as approximations of these milestones.


```python
df_retention_milestones = df_retention_pivot.loc[[3.125, 5.125, 10.125]]

df_retention_milestones.round(2)
```

## Estimated Vehicle Value Comparison

Calculate hypothetical remaining vehicle values at the selected age milestones, assuming an identical original vehicle price of $40,000 across all four powertrain categories.

This comparison illustrates the potential financial implications of differences in value retention.

The estimated values are illustrative and do not represent actual vehicle transaction prices or dealership profitability.


```python
df_estimated_values = 40000 * (df_retention_milestones / 100)

df_estimated_values.round(2)
```

## Visualize Value Retention at Key Vehicle Ages

Create a grouped bar chart to compare median value retention across the four powertrain categories at approximately 3, 5, and 10 years.

This visualization highlights differences in value retention at selected vehicle age milestones.


```python
df_retention_milestones.plot(
    kind="bar",
    figsize=(12, 6),
    title="Used Vehicle Value Retention at Key Ages",
    xlabel="Vehicle Age (Years)",
    ylabel="Value Retention (%)",
    rot=0,
)

plt.grid(axis="y")

plt.savefig(
    "../visualizations/value_retention_milestones.png",
    dpi=300,
    bbox_inches="tight",
)

plt.show()
```

## Key Findings

### 1. BEVs Demonstrate the Greatest Depreciation at Selected Milestones

BEVs exhibit the lowest median value retention across all three selected vehicle age milestones, retaining only 28.02% of their original value at approximately ten years.

### 2. HEVs Demonstrate Strong Early Value Retention

HEVs achieve the highest median value retention at approximately three years (83.68%). However, gasoline vehicles demonstrate slightly stronger value retention at the five-year and ten-year milestones.

### 3. Gasoline and HEV Vehicles Demonstrate Stronger Long-Term Value Retention

At approximately ten years, gasoline and HEV vehicles retain 43.44% and 42.50% of their original value, respectively, outperforming both PHEVs and BEVs.

### 4. PHEVs Underperform Conventional Hybrids

Despite sharing characteristics with conventional hybrid vehicles, PHEVs consistently demonstrate lower value retention than HEVs at the selected age milestones. Further research is required to understand the factors contributing to this difference.

### 5. Depreciation Differences Have Meaningful Financial Implications

Assuming an identical original vehicle price of $40,000, a five-year-old gasoline vehicle retains approximately $6,000 more in estimated value than a comparable BEV under this hypothetical scenario.

This illustrates the potential financial significance of depreciation differences when evaluating used vehicle inventory, but does not establish actual transaction prices or dealership profitability.


## Business Recommendations

### 1. Establish Powertrain-Specific Valuation Guidelines

Dealerships should incorporate vehicle age and powertrain-specific value retention benchmarks into their trade-in and acquisition valuation processes rather than applying uniform depreciation assumptions across all vehicles.

This approach could help dealerships establish more reasonable acquisition offers and reduce the risk of overvaluing or undervaluing used vehicles.

### 2. Incorporate Depreciation into Acquisition Pricing

Rather than prioritizing or excluding specific powertrain categories solely based on depreciation, dealerships should adjust their acquisition offers according to expected resale values, vehicle condition, reconditioning expenses, and target gross profit.

A proposed acquisition pricing framework is:

**Maximum Acquisition Offer = Estimated Resale Value − Reconditioning Costs − Target Gross Profit**

For example, assuming a five-year-old BEV has an estimated resale value of $21,880, requires $2,000 in reconditioning, and has a target gross profit of $3,000, the maximum acquisition offer would be $16,880 before accounting for other applicable costs.

### 3. Strengthen Vehicle Condition and Reconditioning Assessments

Dealerships should carefully assess vehicle condition, mileage, repair requirements, and potential reconditioning expenses before finalizing acquisition offers.

For BEVs and PHEVs, battery health should also be considered when evaluating expected resale value and potential repair expenses.

These assessments, combined with powertrain-specific value retention benchmarks, could help dealerships make more informed acquisition decisions and potentially protect their profit margins.

### Recommendation Summary

The findings support using depreciation and value retention as preliminary valuation benchmarks rather than standalone indicators of inventory profitability.

Dealerships should combine these benchmarks with current market prices, individual vehicle characteristics, and actual operating costs to establish appropriate acquisition offers.


## Limitations and Future Research

### Analysis Limitations

- **Data Granularity:** The dataset contains aggregated value retention statistics rather than individual vehicle records, limiting our ability to evaluate specific vehicles.
- **Vehicle Characteristics:** Individual vehicle condition, mileage, make, model, battery health, and maintenance history were not evaluated. These factors may significantly influence vehicle valuation.
- **Pricing Accuracy:** The analysis is based on listing prices rather than actual transaction prices, which may differ from final negotiated selling prices.
- **Time Coverage:** The aggregated dataset does not allow us to isolate depreciation patterns by calendar year or evaluate changes in market conditions over time.
- **Business Performance:** Actual dealership acquisition costs, reconditioning expenses, customer demand, inventory turnover, and profit margins were not available for analysis.
- **Vehicle Composition:** The analysis compares aggregated vehicle groups at different age intervals rather than tracking individual vehicles over time. Differences in make, model, and vehicle composition may influence observed value retention patterns.

### Future Research Opportunities

Additional data could expand this analysis to support more comprehensive dealership decision-making.

Potential areas for future research include:

- Depreciation patterns by make, model, mileage, and vehicle condition.
- Historical sales volume and customer demand across powertrain categories.
- Inventory availability, acquisition volume, and inventory turnover.
- Actual transaction prices, acquisition costs, and reconditioning expenses.
- Validation of the proposed acquisition pricing framework using actual dealership transactions.

### Conclusion

Despite these limitations, this analysis establishes a foundation for understanding how observed value retention varies across vehicle ages and powertrain categories.

The findings can serve as preliminary valuation benchmarks to support more informed trade-in assessments and inventory acquisition decisions.

As additional data becomes available, this framework can be expanded to support more comprehensive valuation, profitability, and inventory management strategies.
